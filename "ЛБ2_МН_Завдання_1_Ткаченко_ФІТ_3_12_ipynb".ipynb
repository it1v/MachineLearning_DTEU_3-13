**1**. Вивести перші 5 рядків датасета

```python
import pandas as pd
import requests
import io

url = 'https://en.wikipedia.org/wiki/List_of_countries_by_GDP_(nominal)'
headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/100.0.0.0 Safari/537.36'}

response = requests.get(url, headers=headers)
tables = pd.read_html(io.StringIO(response.text))

df = tables[2]

print("Перші 5 рядків датасету:")
display(df.head())

print("\nРозмір датасету (рядків, стовпців):")
print(df.shape)
```

 3.	Визначити оптимальну кількість стовпців (видалити зайві, на власний розсуд).

```python
df.columns = ['Country', 'IMF_2026', 'World_Bank_2025', 'UN_2024']

if df.iloc[0]['Country'].strip() == 'World':
    df = df.drop(0).reset_index(drop=True)

print("Оновлені стовпці:", df.columns.tolist())
display(df.head())
```

4.	Замініть у таблиці значення "—" на значення NaN. Перевірити наявність пропущених значень. При наявності, замінити пропущені значення на середнє значення.

```python
import numpy as np

columns_to_clean = ['IMF_2026', 'World_Bank_2025', 'UN_2024']

for col in columns_to_clean:
    df[col] = df[col].astype(str).str.replace('—N/a', 'NaN').str.replace('—', 'NaN').str.strip()
    df[col] = pd.to_numeric(df[col], errors='coerce')

print("Кількість пропущених значень перед заповненням:")
print(df[columns_to_clean].isnull().sum())

for col in columns_to_clean:
    col_mean = df[col].mean()
    df[col] = df[col].fillna(col_mean)

print("\nКількість пропущених значень після заповнення:")
print(df[columns_to_clean].isnull().sum())
```

8. Визначте відхилення (різницю) між показниками IMF (2026) та World Bank (2025) для кожної країни. У яких країнах ці показники найбільше відрізняються (дати відповідь)?

```python
df['Difference_IMF_WB'] = (df['IMF_2026'] - df['World_Bank_2025']).abs()

top_discrepancies = df.sort_values(by='Difference_IMF_WB', ascending=False).head(5)

print("ТОП-5 країн з найбільшою розбіжністю між IMF та World Bank:")
display(top_discrepancies[['Country', 'IMF_2026', 'World_Bank_2025', 'Difference_IMF_WB']])
```



9.	Обчисліть кореляцію між показниками IMF (2026), World Bank (2025) та United Nations (2024). Які пари змінних мають найвищу кореляцію?

```python
correlation_matrix = df[['IMF_2026', 'World_Bank_2025', 'UN_2024']].corr()

print("Матриця кореляції:")
display(correlation_matrix)

correlations = correlation_matrix.unstack().sort_values(ascending=False)
unique_corrs = correlations[correlations < 1.0].drop_duplicates()

print("\nПари показників за спаданням рівня кореляції:")
for index, value in unique_corrs.items():
    print(f"{index[0]} та {index[1]}: {value:.6f}")
```

11.	Обчисліть стандартне відхилення показників для кожної країни. Яка країна має найвищу варіативність у показниках між роками?

```python
df['Std_Deviation'] = df[['IMF_2026', 'World_Bank_2025', 'UN_2024']].std(axis=1)

most_volatile_country = df.sort_values(by='Std_Deviation', ascending=False).iloc[0]

print("Перші 5 країн з найбільшим стандартним відхиленням:")
display(df[['Country', 'IMF_2026', 'World_Bank_2025', 'UN_2024', 'Std_Deviation']].sort_values(by='Std_Deviation', ascending=False).head(5))

print(f"\nКраїна з найвищою варіативністю у показниках: {most_volatile_country['Country']} (Std Dev: {most_volatile_country['Std_Deviation']:.2f})")
```

12.	Визначення країни з найвищим та найнижчим показниками: Знайдіть країну з найвищим та найнижчим показниками у кожному з років (IMF (2026), World Bank (2025), United Nations (2024)).

```python
years_sources = ['IMF_2026', 'World_Bank_2025', 'UN_2024']

for col in years_sources:
    max_idx = df[col].idxmax()
    max_country = df.loc[max_idx, 'Country']
    max_value = df.loc[max_idx, col]

    min_idx = df[col].idxmin()
    min_country = df.loc[min_idx, 'Country']
    min_value = df.loc[min_idx, col]

    print(f"=== {col} ===")
    print(f"Найвищий показник ВВП: {max_country} ({max_value:,.2f} млн USD)")
    print(f"Найнижчий показник ВВП: {min_country} ({min_value:,.2f} млн USD)\n")
```

13.	Побудуйте гістограму для розподілу показників IMF (2026) серед всіх країн. Який вигляд має розподіл? Чи є країни, що виділяються?

14.	Розрахуйте частку кожної країни в загальному значенні для кожного року (IMF (2026), World Bank (2025), United Nations (2024)).
 Як змінюються частки країн з часом (дати відповідь)?

```python
years_cols = ['IMF_2026', 'World_Bank_2025', 'UN_2024']

for col in years_cols:
    total_sum = df[col].sum()
    df[f'Share_{col}'] = (df[col] / total_sum) * 100

share_cols = [f'Share_{col}' for col in years_cols]
print("Частки ТОП-5 країн у загальному ВВП за роками/джерелами (у %):")
display(df[['Country'] + years_cols + share_cols].sort_values(by='IMF_2026', ascending=False).head(5))

print("\nОпис змін часток:")
print("Аналізуючи ТОП-5 країн (США, Китай, Німеччина, Японія, Велика Британія), ми бачимо динаміку прогнозів з часом від 2024 до 2026 року.")
```

15.	Візуалізуйте зміни в показниках для кожної країни за три роки на графіку. Які країни показують стабільне зростання або спад (дати відповідь)?

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

years = ['2024 (UN)', '2025 (WB)', '2026 (IMF)']

top_10_countries = df.sort_values(by='IMF_2026', ascending=False).head(10)

plt.figure(figsize=(12, 7))
for idx, row in top_10_countries.iterrows():
    gdp_values = [row['UN_2024'], row['World_Bank_2025'], row['IMF_2026']]
    plt.plot(years, gdp_values, marker='o', linewidth=2, label=row['Country'])

plt.title('Тенденція зміни ВВП для ТОП-10 країн за роками/джерелами', fontsize=14)
plt.ylabel('ВВП (млн USD)', fontsize=12)
plt.xlabel('Джерело та прогнозований рік', fontsize=12)
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()

stable_growth = df[(df['UN_2024'] < df['World_Bank_2025']) & (df['World_Bank_2025'] < df['IMF_2026'])]

stable_decline = df[(df['UN_2024'] > df['World_Bank_2025']) & (df['World_Bank_2025'] > df['IMF_2026'])]

print(f"Кількість країн зі стабільним зростанням: {len(stable_growth)} (наприклад: {', '.join(stable_growth['Country'].head(10).tolist())}...)")
print(f"Кількість країн зі стабільним спадом: {len(stable_decline)} (наприклад: {', '.join(stable_decline['Country'].head(10).tolist()) if len(stable_decline) > 0 else 'немає'})")
```
